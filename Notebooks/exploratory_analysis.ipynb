# ☕ Bright Coffee Shop Sales Analysis — Exploratory Analysis
### Purpose
Explore and understand the raw transactional data to uncover patterns, missing values, and initial insights.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

## 📂 Load Dataset

```
data = pd.read_csv('../data/raw/Bright Coffee Shop Analysis - Transactions (7).csv')
data.head()
```

## 🔍 Basic Information

```
data.info()
data.describe(include='all')
```

## 🧹 Check for Missing or Duplicated Values

```
data.isnull().sum()
data.duplicated().sum()
```

## 📊 Exploratory Visualizations

```
# Total transactions per location
sns.countplot(y='store_location', data=data)
plt.title('Transactions by Store Location')
plt.show()

# Distribution of product categories
sns.countplot(y='product_category', data=data)
plt.title('Product Categories Distribution')
plt.show()
```

## 💰 Unit Price Distribution

```
sns.histplot(data['unit_price'], bins=30, kde=True)
plt.title('Distribution of Unit Prices')
plt.show()
```

## 🕒 Transaction Time Patterns

```
data['transaction_time'] = pd.to_datetime(data['transaction_time'], errors='coerce')
data['hour'] = data['transaction_time'].dt.hour
sns.countplot(x='hour', data=data)
plt.title('Transactions by Hour of Day')
plt.show()
```
