# ☕ Bright Coffee Shop Sales Analysis — Final Analysis
### Purpose
Analyze sales performance, revenue drivers, and product trends to provide data-driven recommendations.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv('../data/processed/coffee_shop_cleaned.csv')
data.head()
```

## 💰 Compute Total Revenue

```
data['total_amount'] = data['unit_price'] * data['transaction_qty']
data.head()
```

## 📈 Total Revenue by Product Type

```
revenue_by_type = data.groupby('product_type')['total_amount'].sum().sort_values(ascending=False)
revenue_by_type.plot(kind='bar', figsize=(10,5), title='Revenue by Product Type')
plt.show()
```

## 🕒 Peak Sales Hours

```
data['transaction_time'] = pd.to_datetime(data['transaction_time'], errors='coerce')
data['hour'] = data['transaction_time'].dt.hour
hourly_sales = data.groupby('hour')['total_amount'].sum()
hourly_sales.plot(kind='line', figsize=(10,5), title='Hourly Sales Trend')
plt.show()
```

## 🏬 Store Location Performance

```
store_sales = data.groupby('store_location')['total_amount'].sum().sort_values(ascending=False)
sns.barplot(x=store_sales.values, y=store_sales.index)
plt.title('Total Revenue by Store Location')
plt.show()
```

## 🧁 Top 10 Best-Selling Products

```
top_products = data.groupby('product_detail')['total_amount'].sum().nlargest(10)
sns.barplot(x=top_products.values, y=top_products.index)
plt.title('Top 10 Products by Revenue')
plt.show()
```
